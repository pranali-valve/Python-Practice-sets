# Built-in Functions in tuple

```python
t1=(3,5,2,2,2,2,6,5,8)
print(t1)
```

```python
#1. count()= count repeated element
t1.count(2)
```

```python
#2. sum()= Addition of whole tuple
sum(t1)
```

```python
#3. len()= Define length
len(t1)
```

```python
#4. min()= Find minimum value
min(t1)
```

```python
#5. max()= Find maximum value
max(t1)
```
