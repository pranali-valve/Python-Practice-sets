# Series create

```python
# creating a series from a list
arr=[0,1,2,3,4]
s1=pd.Series(arr)
s1
```

# Series Operations

```python
arr1=[0,1,2,3,4,5,7]
arr2=[6,7,8,9,5]
```

```python
s5=pd.Series(arr2)
s5
```

```python
s5.add(s6)
```

```python
s5.sub(s6)
```
