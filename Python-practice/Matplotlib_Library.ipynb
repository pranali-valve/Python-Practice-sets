# Matplotlib Library

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.plot(x,y)
plt.xlabel("Month")
plt.ylabel("Number")
```

# remove heading

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.plot(x,y)
plt.show()
```

# Change line color

```python
x = [1,2,3,4]
y = [5,6,7,8]
c='red'
plt.plot(x,y,c)
plt.show()
```

# Use Marker

```python
x = [1,2,3,4]
y = [5,6,7,8]
c='r'
plt.plot(x,y,c,marker='o')
plt.show()
```

# Show different plot

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.plot(x,y)
plt.plot(y,marker='>')
plt.show()
```

# Bar graph

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.bar(x,y)
plt.show()
```

# Change bar graph color 

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = ['r','g','b','y']
plt.bar(x,y, color = c)
plt.show()
```

# Scatter Plot

```python
x=[1,2,3,4,5,6]
y=[5,6,7,8,9,2]
plt.title("scatter plot")
plt.scatter(x,y)
plt.show()
```

# Adding Label

```python
x=[1,2,3,4,5,6]
y=[5,6,7,8,9,2]
plt.title("scatter plot")
plt.xlabel("Month")
plt.ylabel("Number")
plt.scatter(x,y)
plt.show()
```

# Change Color

```python
x=[1,2,3,4,5,6]
y=[5,6,7,8,9,2]
plt.title("scatter plot")
plt.xlabel("Month")
plt.ylabel("Number")
c=['red','magenta','green','orange','blue','brown']
plt.scatter(x,y,color=c)
plt.show()
```

# Change size

```python
x=[1,2,3,4,5,6]
y=[5,6,7,8,9,2]
plt.title("scatter plot")
plt.xlabel("Month")
plt.ylabel("Number")
c=['red','magenta','green','orange','blue','brown']
plt.scatter(x,y,color=c, s=150)
plt.show()
```

# Add Image

# Image Title

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
fname = r'flower.jpeg'
image = Image.open(fname).convert('RGB')
plt.imshow(image, cmap = 'gray')
plt.title("Frangipani Flower")
plt.show()
```

# Xlabel & Ylabel

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
fname = r'flower.jpeg'
image = Image.open(fname).convert('RGB')
plt.imshow(image, cmap = 'gray')
plt.title("Frangipani Flower")
plt.xlabel("Width")
plt.ylabel("Height")
plt.show()
```

# Grid

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
fname = r'flower.jpeg'
image = Image.open(fname).convert('RGB')
plt.imshow(image, cmap = 'gray')
plt.title("Frangipani Flower")
plt.xlabel("Width")
plt.ylabel("Height")
plt.grid()
plt.show()
```

# Pie Chart

```python
import matplotlib.pyplot as plt
```

```python
a=[10,20,30,40]
b=["Java","C++","DBMS","Python"]
plt.pie(a, labels=b)
plt.show()
```

# Change color

```python
a=[10,20,30,40]
b=["Java","C++","DBMS","Python"]
c = ["Yellow","magenta","Brown","aqua"]
plt.pie(a, labels=b, colors = c)
plt.show()
```

# legend

```python
a=[10,20,30,40]
b=["Java","C++","DBMS","Python"]
c = ["Yellow","magenta","Brown","aqua"]
plt.pie(a, labels=b, colors = c)
plt.legend()
plt.show()
```
