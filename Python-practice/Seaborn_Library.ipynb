# Using Seaborn 

```python
var = [1,2,3,4,5,6,7,8]
var1 = [8,7,6,5,4,3,2,1]
sns.lineplot(x = var, y = var1)
plt.show()
```

# To create dataframe plot using seaborn library

```python
var = [1,2,3,4,5,6,7,8]
var1 = [8,7,6,5,4,3,2,1]
import pandas as pd
df = pd.DataFrame({"var":var,"var1":var1})
sns.lineplot(x = 'var', y = 'var1', data = df)
plt.show()
```

```python
df
```

# Load Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = sns.load_dataset("iris")
x
```

# Draw Lineplot

```python
a1 = sns.lineplot(x="sepal_length", y="sepal_width", data=x)
plt.show()
```

# Dividing Plots in Categories using "Hue"

```python
a2 = sns.lineplot(x="sepal_length", y="sepal_width", data=x, hue="species")
plt.show()
```

# Change Plot Color

```python
a3 = sns.lineplot(x="sepal_length", y="sepal_width", data=x, hue="species", palette="ocean")
plt.show()
```

# Use Markers

```python
a3 = sns.lineplot(x="sepal_length", y="sepal_width", data=x, hue="species", palette="ocean", style="species", markers=["o",">","<"])
plt.show()
```

```python
a4 = sns.lineplot(x="sepal_length", y="sepal_width", data=x, hue="species", palette="ocean", style="species", markers=["o",">","<"])
plt.grid()
plt.show()
```
